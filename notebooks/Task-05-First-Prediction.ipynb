```python
import os
import sys

project_root = os.path.abspath("..")

if project_root not in sys.path:
    sys.path.append(project_root)

print("Project Root:", project_root)
```

```python
import pandas as pd

from src.data import load_data, split_data
from src.features import (
    build_preprocessor,
    fit_preprocessor,
    transform_data,
)
from src.model import (
    build_baseline,
    build_model,
    train_model,
)
from src.evaluate import (
    evaluate_model,
    detailed_evaluation,
)
```

```python
df = load_data()

print("Dataset Shape:", df.shape)

df.head()
```

```python
X_train, X_val, X_test, y_train, y_val, y_test = split_data(df)

print("Training:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)
```

```python
numerical_features = X_train.columns.tolist()

preprocessor = build_preprocessor(numerical_features)

preprocessor = fit_preprocessor(preprocessor, X_train)

print("Preprocessor Ready!")
```

```python
X_train = transform_data(preprocessor, X_train)
X_val = transform_data(preprocessor, X_val)
X_test = transform_data(preprocessor, X_test)

print("Data transformed successfully!")
```

```python
baseline = build_baseline()

baseline = train_model(
    baseline,
    X_train,
    y_train
)

baseline_accuracy = evaluate_model(
    baseline,
    X_val,
    y_val
)

print(f"Baseline Accuracy: {baseline_accuracy:.4f}")
```

```python
model = build_model()

model = train_model(
    model,
    X_train,
    y_train
)

model_accuracy = evaluate_model(
    model,
    X_val,
    y_val
)

print(f"Logistic Regression Accuracy: {model_accuracy:.4f}")
```

```python
comparison = pd.DataFrame({
    "Model": [
        "DummyClassifier",
        "LogisticRegression"
    ],
    "Accuracy": [
        baseline_accuracy,
        model_accuracy
    ]
})

comparison
```

```python
predictions, cm, report = detailed_evaluation(
    model,
    X_val,
    y_val
)

print("Confusion Matrix:\n")

print(cm)
```
