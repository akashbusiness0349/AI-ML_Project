```python
import os
import sys

project_root = os.path.abspath("..")

if project_root not in sys.path:
    sys.path.append(project_root)

print("Project Root:", project_root)
```

```python
from src.data import load_data, split_data
from src.features import (
    build_preprocessor,
    fit_preprocessor,
    transform_data,
    save_preprocessor,
)

from src.config import CONFIG

import pandas as pd
```

```python
df = load_data()

print("Dataset Shape:", df.shape)

df.head()
```

```python
X_train, X_val, X_test, y_train, y_val, y_test = split_data(df)

print("Training Set :", X_train.shape)
print("Validation Set:", X_val.shape)
print("Test Set :", X_test.shape)
```

```python
numerical_features = X_train.columns.tolist()

print("Numerical Features:")

numerical_features
```

```python
preprocessor = build_preprocessor(numerical_features)

print(preprocessor)
```

```python
preprocessor = fit_preprocessor(preprocessor, X_train)

print("Preprocessor fitted successfully!")
```

```python
X_train_processed = transform_data(preprocessor, X_train)

X_val_processed = transform_data(preprocessor, X_val)

X_test_processed = transform_data(preprocessor, X_test)

print("Training Shape:", X_train_processed.shape)
print("Validation Shape:", X_val_processed.shape)
print("Test Shape:", X_test_processed.shape)
```

```python
pd.DataFrame(
    X_train_processed,
    columns=numerical_features
).head()
```

```python
import numpy as np

print("Missing Values in Train:", np.isnan(X_train_processed).sum())
print("Missing Values in Validation:", np.isnan(X_val_processed).sum())
print("Missing Values in Test:", np.isnan(X_test_processed).sum())
```

```python
import os

os.makedirs("../models", exist_ok=True)

save_preprocessor(
    preprocessor,
    "../models/preprocessor.pkl"
)

print("Preprocessor saved successfully!")
```
