**Objective:**
Understand the machine learning problem by defining the target, identifying useful features, checking data quality, detecting leakage, and deciding the appropriate modelling approach before training any advanced model.

```python
import pandas as pd
import numpy as np

from sklearn.datasets import load_iris
```

```python
# Load the Iris dataset
iris = load_iris()

# Create a DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Add the target column
df["target"] = iris.target

# Display the first five rows
df.head()
```

### Problem Statement

Predict the species of an Iris flower using its sepal length, sepal width, petal length, and petal width measurements.

### Success Metric

The model will be evaluated using **Accuracy**, because the Iris dataset is a balanced multi-class classification dataset.

## Target Definition

The target variable is **target**, which represents the species of the Iris flower.

Target values:

- 0 → Setosa
- 1 → Versicolor
- 2 → Virginica

The objective of the machine learning model is to correctly predict the target class using the flower measurements.

```python
# Display unique target values
print("Unique Target Values:", df["target"].unique())

# Display class names
print("Class Names:", iris.target_names)
```

```python
# Display information about the dataset
df.info()
```

```python
# Statistical summary of numerical features
df.describe()
```

```python
# Check for missing values
print(df.isnull().sum())
```

```python
# Check class distribution
class_counts = df["target"].value_counts()

print(class_counts)
```

## Modelling Decision

### Decision

- **Problem Type:** Multi-class Classification
- **Target Variable:** target
- **Evaluation Metric:** Accuracy
- **Dataset Balance:** Balanced
- **Leakage:** Not Detected

### Go / No-Go Decision

**GO**

The dataset contains meaningful features, no leakage, balanced classes, and is suitable for training a classification model.
